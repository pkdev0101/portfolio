## Challenge: Find the Bug

The toggle was missing its track class, so there was nothing visible to click. Adding `ocs__toggle-track` to the empty `span` fixes it. Click **Run** to see "Toggle works! 🎉".

```python
%%html
<!-- UI_RUNNER: Challenge - Find the Bug | interactive: true, render_mode: javascript -->
<div id="toggles-bug-root"></div>
<script>
// Falls back to the div above when this cell runs in Jupyter instead of the UI runner.
const out = (typeof outputElement !== "undefined") ? outputElement : document.getElementById("toggles-bug-root");

out.innerHTML = [
  '<label class="ocs__toggle ocs__toggle--green">',
  '  <input class="ocs__toggle-input" type="checkbox">',
  // The fix: the track is the visible switch. aria-hidden tells screen readers to skip it
  // because it's only decoration; the label text already describes the toggle.
  '  <span class="ocs__toggle-track" aria-hidden="true"></span>',
  '  <span class="ocs__toggle-label">Enable Notifications</span>',
  '</label>'
].join('');

const result = document.createElement("p");
result.textContent = out.querySelector(".ocs__toggle-track") ? "Toggle works! 🎉" : "Incorrect, almost there!";
out.appendChild(result);
</script>
```

## Challenge: Customize the Toggle

I changed the checked color from green (`#22C55E`) to purple (`#8B5CF6`). Click **Run**, then flip the toggle to see the new color.

```python
%%html
<!-- UI_RUNNER: Challenge - Customize the Toggle | interactive: true, render_mode: javascript -->
<div id="toggles-color-root"></div>
<script>
const out = (typeof outputElement !== "undefined") ? outputElement : document.getElementById("toggles-color-root");

out.innerHTML = [
  '<style>',
  // ":checked + .ocs__toggle-track" means "the track right after a checked input",
  // so the new color only shows when the toggle is ON.
  '.custom .ocs__toggle-input:checked + .ocs__toggle-track {',
  '  background: #8B5CF6;',
  '}',
  '</style>',
  '<div class="custom">',
  '  <label class="ocs__toggle">',
  '    <input class="ocs__toggle-input" type="checkbox">',
  '    <span class="ocs__toggle-track" aria-hidden="true"></span>',
  '    <span class="ocs__toggle-label">My Custom Purple Toggle</span>',
  '  </label>',
  '</div>'
].join('');
</script>
```

## Homework: Study Session Settings

Click **Run** in the UI runner, then flip the switches. This panel has 4 toggles built with the `ocs__toggle` SASS classes, and each one controls a real feature:

- **Night Mode** changes the panel's colors.
- **Show Study Tip** shows or hides a study tip.
- **Large Text** makes the panel's text bigger.
- **Focus Mode** hides the distractions section.

A live counter shows how many toggles are on, and a bonus message appears when all four are on. The comments in the code explain how each part works.

```python
%%html
<!-- UI_RUNNER: Homework - Study Session Settings | interactive: true, render_mode: javascript -->
<div id="toggles-hw-root"></div>
<script>
// In the UI runner, outputElement is the output box. In Jupyter it doesn't exist,
// so the code falls back to the div above. This lets the same code run in both places.
const root = (typeof outputElement !== "undefined") ? outputElement : document.getElementById("toggles-hw-root");

// Each toggle uses the 4-part OCS structure: label > input > track > label text.
// The input is hidden by the SCSS but still stores the on/off state.
function makeToggle(id, text) {
  return '<label class="ocs__toggle ocs__toggle--green">' +
    '<input id="' + id + '" class="ocs__toggle-input" type="checkbox">' +
    '<span class="ocs__toggle-track" aria-hidden="true"></span>' +
    '<span class="ocs__toggle-label">' + text + '</span>' +
    '</label><br>';
}

root.innerHTML =
  '<div id="study-panel" class="ocs__card">' +
    '<h3>Study Session Settings</h3>' +
    makeToggle("night-toggle", "Night Mode") +
    makeToggle("tip-toggle", "Show Study Tip") +
    makeToggle("text-toggle", "Large Text") +
    makeToggle("focus-toggle", "Focus Mode (hide distractions)") +
    '<p id="study-tip" hidden><strong>Tip:</strong> Study in 25-minute blocks with 5-minute breaks (the Pomodoro Technique).</p>' +
    '<div id="distractions"><strong>Distractions:</strong> 3 new messages, group chat active, new video uploaded</div>' +
    // aria-live="polite" makes screen readers announce the count when it changes.
    '<p><output id="toggle-counter" aria-live="polite">Toggles on: 0 of 4</output></p>' +
    '<p id="all-on-message" hidden><strong>Full study mode unlocked!</strong></p>' +
  '</div>';

// I search inside root instead of the whole page so these IDs can't clash with other parts of my site.
const panel = root.querySelector("#study-panel");
const nightToggle = root.querySelector("#night-toggle");
const tipToggle = root.querySelector("#tip-toggle");
const textToggle = root.querySelector("#text-toggle");
const focusToggle = root.querySelector("#focus-toggle");
const allToggles = root.querySelectorAll(".ocs__toggle-input");

function updatePage() {
  // .checked is a Boolean, so it works directly as a condition.
  // This is AP CSP "selection": the program picks a behavior based on true or false.

  // Feature 1: Night Mode changes the panel's colors.
  panel.style.background = nightToggle.checked ? "#1e293b" : "";
  panel.style.color = nightToggle.checked ? "#e2e8f0" : "";

  // Feature 2: The "!" flips the value. Toggle ON means hidden = false, so the tip shows.
  root.querySelector("#study-tip").hidden = !tipToggle.checked;

  // Feature 3: Large Text makes everything in the panel bigger.
  panel.style.fontSize = textToggle.checked ? "1.3em" : "";

  // Feature 4: Focus Mode is the opposite of the tip. ON should HIDE the section, so no "!".
  root.querySelector("#distractions").hidden = focusToggle.checked;

  // Counter: count resets to 0 every time, so it always shows the CURRENT state
  // instead of adding up forever.
  let count = 0;
  allToggles.forEach(function (toggle) {
    if (toggle.checked) {
      count++;
    }
  });
  root.querySelector("#toggle-counter").textContent = "Toggles on: " + count + " of " + allToggles.length;

  // Comparing to allToggles.length instead of typing 4 means a 5th toggle would still work.
  root.querySelector("#all-on-message").hidden = count !== allToggles.length;
}

// I used "change" instead of "click" because it fires whenever the checked state changes,
// including keyboard use (Tab + Space), not just mouse clicks.
allToggles.forEach(function (toggle) {
  toggle.addEventListener("change", updatePage);
});

// Run once at the start so the counter and features match the starting state.
updatePage();
</script>
```

## How It Works

The hidden checkbox inside each `ocs__toggle` stores the on/off state, and the SASS `:checked` styling changes how the switch looks. JavaScript reads that same state with `.checked`, which is a true/false value, and uses it as the condition for each feature. This is selection from AP CSP: the program chooses which behavior to run based on whether a condition is true. One shared `updatePage()` function handles every toggle, so adding another toggle only takes one more line of HTML.

## Popcorn Hack

Click **Run** in the UI runner to render the refactored inputs. The inline `width` and dashed `border` styles and the made-up class are replaced with the OCS input grammar: one large gradient search field and one compact plain field.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack - PVO Search Inputs | interactive: true -->
<!-- large + gradient replaces the inline 400px width and dashed purple border. -->
<label for="pvo-search">Search PVO resources</label>
<input id="pvo-search" type="text" class="ocs__input large gradient" placeholder="Search PVO resources...">
<!-- small replaces the made-up class from the starter with a compact plain field. -->
<label for="volunteer-first-name">Volunteer first name</label>
<input id="volunteer-first-name" type="text" class="ocs__input small" placeholder="Volunteer First Name">
```

## Homework Hack

Click **Run** in the UI runner to render the refactored PVO signup form. Every input uses `ocs__input` with a size modifier, the gradient field uses the `gradient` modifier instead of an inline background, and every field has a visible label plus a placeholder.

```python
%%html
<!-- UI_RUNNER: Homework Hack - PVO Signup Form | interactive: true -->
<!-- Each label's "for" matches its input's id, so clicking the label focuses the field
     and screen readers announce what the field is for. -->
<label for="assistance-category">Assistance category</label>
<!-- The gradient modifier replaces the inline linear-gradient background, and large replaces width: 500px. -->
<input id="assistance-category" type="text" class="ocs__input large gradient" placeholder="Assistance Category">

<label for="signup-first-name">Volunteer first name</label>
<!-- small replaces the made-up class from the starter. A first name is short, so a compact field fits. -->
<input id="signup-first-name" type="text" class="ocs__input small" placeholder="Volunteer First Name">

<label for="signup-email">Volunteer email</label>
<!-- medium replaces the inline padding. type="email" stays because it fits the data and
     gives phones an email keyboard. -->
<input id="signup-email" type="email" class="ocs__input medium" placeholder="Volunteer Email">
```

## How It Works

HTML defines each input and its label, the OCS SASS grammar styles it through `ocs__input` plus a size or style modifier, and no inline CSS is needed. Using the same base class everywhere keeps the PVO form consistent, so a visitor doesn't have to relearn the interface from field to field. In AP CSP pseudocode, each field is like an `INPUT()` call: the label and placeholder tell the user what data the program is asking for.

## Popcorn Hack 1: Add the Base Button Class

Click **Run** to render the button. Adding `ocs__btn` gives the link the shared OCS button styling with no CSS.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 - Base Button Class | interactive: true -->
<!-- ocs__btn is the base class. An <a> is used because this navigates to another page. -->
<a class="ocs__btn" href="https://www.mygoodbrain.org/resources">Resource Directory</a>
```

## Popcorn Hack 2: Modifiers and Button Groups

Click **Run** to render the group. The check-in button gets `pill` and `fill`, and both actions are wrapped in an `ocs__links` container for spacing.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 - Modifiers and Button Groups | interactive: true -->
<!-- ocs__links groups the actions and handles spacing, so no flex or gap CSS is needed. -->
<div class="ocs__links">
  <!-- pill rounds the corners and fill gives it a solid background, marking it as the main action. -->
  <button class="ocs__btn pill fill">Complete Daily Check-In</button>
  <button class="ocs__btn outline">Save Draft Journal</button>
</div>
```

## Popcorn Hack 3: Status Tones and Wide Layout Bar

Click **Run** to render the toolbar. The container is upgraded to `ocs__links--wide`, the second link's tone changes from yellow to `alert-green`, and the crisis button gets `fill` for full visual priority.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 - Status Tones and Wide Layout Bar | interactive: true -->
<!-- ocs__links--wide spreads the actions across a wider responsive toolbar. -->
<div class="ocs__links--wide">
  <button class="ocs__btn alert-green fill">Complete A Mental Health Check-In Form</button>
  <!-- alert-green replaces alert-yellow, so this action now reads as positive instead of cautionary. -->
  <a class="ocs__btn alert-green outline" href="https://www.mygoodbrain.org/resources">Request Break</a>
  <!-- fill on the red tone gives the most urgent action the strongest visual weight. -->
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources">Get Crisis Support</a>
</div>
```

## Homework Hack 1: Adaptive Resource Toolbar

Click **Run** to render the toolbar. The inline-styled badges are replaced with OCS status actions that match each category's urgency, all wrapped in `ocs__links`.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 - Adaptive Resource Toolbar | interactive: true -->
<!-- ocs__links replaces the inline padding, display: flex, and gap on the old wrapper. -->
<div class="ocs__links">
  <!-- Red tone = urgent. pill replaces the inline border-radius. -->
  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources">Urgent Support</a>
  <!-- Green tone = positive self-care, replacing the hardcoded teal background. -->
  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources">Self-Care</a>
  <!-- accent outline replaces the hardcoded purple border and text color. -->
  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources">Community</a>
</div>
```

## Homework Hack 2: Primary vs. Secondary Action Priority

Click **Run** to render the action row. Visual hierarchy now comes from OCS modifiers: `fill` for the primary action, `outline` for the secondary action, and an `alert-red outline` for the alert action.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 - Action Priority | interactive: true -->
<!-- ocs__links replaces the inline white background, padding, and border-radius on the old panel. -->
<div class="ocs__links">
  <!-- Primary: a <button> because it performs an action on this page. accent fill = highest priority. -->
  <button class="ocs__btn accent fill">Log Mood Entry</button>
  <!-- Secondary: an <a> because it navigates. outline = medium priority. -->
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">View Weekly Trends</a>
  <!-- Alert: the red tone flags it as important, but outline keeps it from competing with the primary action. -->
  <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources">Crisis Hotline</a>
</div>
```

## How It Works

OCS button grammar works like a procedure with parameters, which connects to AP CSP AAP-3.A: `ocs__btn` is the reusable base, and modifiers like `fill`, `outline`, `pill`, `accent`, and the `alert-*` tones change its appearance without any new CSS. Choosing `<button>` for on-page actions and `<a>` for navigation keeps the HTML semantic, and `ocs__links` handles spacing for the whole group. Picking the tone and fill based on urgency is selection: the more urgent the action, the stronger its visual weight.

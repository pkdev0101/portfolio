## Homework Hack

Refactored "How to Pat a Cat": removed the `<style>` block and custom classes, used correct heading hierarchy, turned the fake numbered lines into one `<ol>` with three `<li>` items, and changed the visual emphasis to `<strong>`.

```python
%%html
<!-- UI_RUNNER: Typography Homework Base -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3>Have you ever wanted to...</h3>
  <p class="ocs_text">pat a cat? Well good for you, we will give step by step instructions on how to pat a cat!</p>
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <ol>
    <li>Put out your hand</li>
    <li>Touch the cat</li>
    <li>Move your hand back and forth</li>
  </ol>
  <h3>Congrats! You pat a cat :D</h3>
</div>
```

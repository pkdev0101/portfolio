## Homework Hack

Click **Run** in the UI runner to render the finished cat instructions as a heading, paragraphs, and a numbered list.

```python
%%html
<!-- UI_RUNNER: Homework Hack — How to Pat a Cat | interactive: true -->
<!-- The card groups the instructions into one readable block. -->
<!-- Headings organize sections and the ordered list presents the steps in sequence. -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3>Have you ever wanted to…</h3>
  <p class="ocs_description">pat a cat? Well good for you we will give step by step instructions on how to pat a cat!</p>
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <ol>
    <li>put out your hand</li>
    <li>touch the cat</li>
    <li>move your hand back and forth</li>
  </ol>
  <h3>Congrats! You pat a cat :D</h3>
</div>
```

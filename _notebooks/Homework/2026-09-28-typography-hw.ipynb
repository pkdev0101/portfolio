## Popcorn Hack

Click **Run** in the UI runner to render the heading, subheading, and paragraph below.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack - Why Cats Purr | interactive: true -->
<!-- h2 is the main heading, h3 is the supporting subheading, and p holds the body text. -->
<!-- OCS classes give each element its role, so no custom CSS is needed. -->
<h2 class="ocs_section-title">Why Cats Purr</h2>
<h3 class="ocs_lead">A Sign of Happiness</h3>
<p class="ocs_text">Cats usually purr when they feel safe, cozy, and ready for a nap.</p>
```

## Homework Hack

Click **Run** in the UI runner to render the finished cat instructions as a heading, paragraphs, and a numbered list.

```python
%%html
<!-- UI_RUNNER: Homework Hack - How to Pat a Cat | interactive: true -->
<!-- ocs_card frames the whole section as one content panel. -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3>Have you ever wanted to...</h3>
  <p class="ocs_text">pat a cat? Well good for you, we will give step by step instructions on how to pat a cat!</p>
  <!-- strong marks real importance instead of just looking bold. -->
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <!-- One ol with three li items; the browser numbers them automatically. -->
  <ol>
    <li>Put out your hand</li>
    <li>Touch the cat</li>
    <li>Move your hand back and forth</li>
  </ol>
  <h3>Congrats! You pat a cat :D</h3>
</div>
```

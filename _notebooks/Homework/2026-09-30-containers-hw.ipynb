## Popcorn Hack: Parade Invite Rescue

Click **Run** in the UI runner to render the rescued invite. The `page-wrap` div, the purple inline title, and the hand-made flex row are replaced with one `ocs__container`, one `ocs__card`, and a 3-column standard grid with one accent cell. There are no inline styles or made-up classes.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack - Parade Invite Rescue | interactive: true -->
<!-- One ocs__container replaces page-wrap and its inline max-width and margin. -->
<div class="ocs__container">
  <!-- The card groups the whole invite into one framed block. -->
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Invite</h2>
    <p class="ocs__description">Join the Poway Scripps Rotary Club for a morning of floats, music, and community.</p>
    <!-- The standard grid replaces style="display: flex" and collapses to fewer columns on phones. -->
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <!-- The accent modifier highlights the most important detail. -->
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start time: 9:00AM</div>
      <div class="ocs__grid-cell">What you get: An invaluable local and cultural experience!</div>
    </div>
  </div>
</div>
```

## Homework Hack: Parade Hub Page

Click **Run** in the UI runner to render the rebuilt Parade hub. It uses exactly one `ocs__container` holding one `ocs__card`, a section title with a one-line description, a 3-column standard grid with a header, accent, and muted cell, a table built with `ocs__table-wrap` and `ocs__table`, and a closing callout. Resize the window to check that the grid collapses cleanly on narrow screens.

```python
%%html
<!-- UI_RUNNER: Homework Hack - Poway Rotary Parade Hub | interactive: true -->
<!-- Exactly one container sets the page boundary for everything below. -->
<div class="ocs__container">
  <!-- Exactly one card holds all of the hub content. -->
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade</h2>
    <p class="ocs__description">Everything you need to know about the Poway Scripps Rotary Club's annual parade.</p>

    <!-- cols-3 standard grid with one header, one accent, and one muted cell. -->
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start time: 9:00AM</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Parade floats from local schools and clubs</div>
    </div>

    <!-- The wrap keeps the table aligned with the container; the table gets the shared OCS style. -->
    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <tr><th>Entry</th><th>Status</th></tr>
        <tr><td>Parade floats</td><td>Open</td></tr>
        <tr><td>Volunteer sign-up</td><td>Confirmed</td></tr>
      </table>
    </div>

    <!-- The callout highlights the key takeaway at the end. -->
    <div class="ocs__callout">One container holds the whole Parade hub, so it looks the same on phones and computers with no custom CSS.</div>
  </div>
</div>
```

## Containers Explanation

The `ocs__container` sets one centered page boundary, and every other OCS block nests inside it: the card groups related content, the grid splits it into responsive columns, and the table and callout reuse shared styles. Because the global SASS handles widths, spacing, and colors, the page needs no inline styles or invented classes, and it adapts to phone and computer screens automatically. This is the same idea as abstraction in AP CSP: the container hides the layout details so I can focus on the content.

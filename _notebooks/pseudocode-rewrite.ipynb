## 4. Procedure with a Return Value

Turn a confirmation count into a status word.

### JavaScript

```python
%%js

// CODE_RUNNER: Add a Deep level for 100 or more.

function getTxStatus(confirmations) {
  if (confirmations >= 6) {
    return "Confirmed";
  } else {
    if (confirmations >= 1) {
      return "Pending";
    } else {
      return "Unconfirmed";
    }
  }
}

let confirmations = 3;
let status = getTxStatus(confirmations);
console.log(confirmations + " confirmations: " + status);
```

### Python

## 7. Iteration with a Sentinel Loop

Keep accepting transactions into the mempool until the miner types done.

### JavaScript

```python
%%js

// CODE_RUNNER: Add one more transaction to the answers list.

let answers = ["tx_9f2a", "1200", "next", "tx_4c81", "3400", "next", "tx_7de0", "850", "done"];
function ask(q) { let r = answers.shift(); console.log(q + " " + r); return r; }

console.log("Mempool");

let mempool = [];
let txCount = 0;
let feeTotal = 0;
let action = "next";

while (action !== "done") {
  let txid = ask("Transaction id:");
  let fee = Number(ask("Fee in sats:"));

  mempool.push(txid);
  txCount = txCount + 1;
  feeTotal = feeTotal + fee;

  action = ask("Type next or done:");
}

console.log("Mempool: " + mempool);
console.log("Transactions: " + txCount);
console.log("Fees waiting: " + feeTotal + " sats");
```

### Python

```python
# CODE_RUNNER: Add one more transaction to the answers list.

answers = ["tx_9f2a", "1200", "next", "tx_4c81", "3400", "next", "tx_7de0", "850", "done"]

def ask(question):
    reply = answers.pop(0)
    print(question + " " + reply)
    return reply

print("Mempool")

mempool = []
tx_count = 0
fee_total = 0
action = "next"

while action != "done":
    txid = ask("Transaction id:")
    fee = int(ask("Fee in sats:"))

    mempool.append(txid)
    tx_count = tx_count + 1
    fee_total = fee_total + fee

    action = ask("Type next or done:")

print("Mempool: " + str(mempool))
print("Transactions: " + str(tx_count))
print("Fees waiting: " + str(fee_total) + " sats")
```

## 8. Algorithm, All Three Pieces

Loop the fees, guard against an empty block, then judge congestion.

### JavaScript

```python
%%js

// CODE_RUNNER: Empty the list. The guard should return 0, not crash.

function averageFee(fees) {
  let total = 0;
  let count = 0;

  for (let fee of fees) {
    total = total + fee;
    count = count + 1;
  }

  if (count > 0) {
    return total / count;
  } else {
    return 0;
  }
}

let blockFees = [1200, 3400, 850, 2100, 1450];
let result = averageFee(blockFees);

console.log("Fees: " + blockFees);
console.log("Average fee: " + result + " sats");

if (result >= 2000) {
  console.log("Network: busy");
} else {
  console.log("Network: calm");
}
```

### Python

```python
# CODE_RUNNER: Empty the list. The guard should return 0, not crash.

def average_fee(fees):
    total = 0
    count = 0

    for fee in fees:
        total = total + fee
        count = count + 1

    if count > 0:
        return total / count
    else:
        return 0

block_fees = [1200, 3400, 850, 2100, 1450]
result = average_fee(block_fees)

print("Fees: " + str(block_fees))
print("Average fee: " + str(result) + " sats")

if result >= 2000:
    print("Network: busy")
else:
    print("Network: calm")
```

## 9. Linear Search

Look for a transaction id inside a block. Missing means -1.

### JavaScript

```python
%%js

// CODE_RUNNER: Search for a txid that is not there.

function findTx(block, target) {
  let index = 1;

  for (let txid of block) {
    if (txid === target) {
      return index;
    }
    index = index + 1;
  }

  return -1;
}

let block = ["tx_9f2a", "tx_4c81", "tx_7de0", "tx_1b55"];

let wanted = "tx_7de0";
let position = findTx(block, wanted);

if (position > 0) {
  console.log(wanted + " is at position " + position);
} else {
  console.log("Not in this block");
}
```

### Python

```python
# CODE_RUNNER: Search for a txid that is not there.

def find_tx(block, target):
    index = 1

    for txid in block:
        if txid == target:
            return index
        index = index + 1

    return -1

block = ["tx_9f2a", "tx_4c81", "tx_7de0", "tx_1b55"]

wanted = "tx_7de0"
position = find_tx(block, wanted)

if position > 0:
    print(wanted + " is at position " + str(position))
else:
    print("Not in this block")
```

## 10. List Operations

Reorder the mempool queue. Pseudocode position 2 is real index 1.

### JavaScript

```python
%%js

// CODE_RUNNER: Insert a transaction at position 1 and watch the rest shift.

let queue = ["tx_9f2a", "tx_4c81"];
console.log("Initial: " + queue);

queue.push("tx_7de0");
console.log("After APPEND: " + queue);

queue.splice(1, 0, "tx_high_fee");   // INSERT at position 2
console.log("After INSERT at 2: " + queue);

queue.splice(2, 1);                  // REMOVE at position 3
console.log("After REMOVE at 3: " + queue);

let waiting = queue.length;
console.log("Waiting: " + waiting);

for (let txid of queue) {
  console.log("Mining: " + txid);
}
```

### Python

```python
# CODE_RUNNER: Insert a transaction at position 1 and watch the rest shift.

queue = ["tx_9f2a", "tx_4c81"]
print("Initial: " + str(queue))

queue.append("tx_7de0")
print("After APPEND: " + str(queue))

queue.insert(1, "tx_high_fee")       # INSERT at position 2
print("After INSERT at 2: " + str(queue))

queue.pop(2)                         # REMOVE at position 3
print("After REMOVE at 3: " + str(queue))

waiting = len(queue)
print("Waiting: " + str(waiting))

for txid in queue:
    print("Mining: " + txid)
```

## 11. Parallel Lists, Find the Position

Two lists line up. Search the addresses, use the position to read the owner.

### JavaScript

```python
%%js

// CODE_RUNNER: Look up bc1_ccc instead.

function findAddress(addressList, target) {
  let index = 1;

  for (let address of addressList) {
    if (address === target) {
      return index;
    }
    index = index + 1;
  }

  return -1;
}

let addresses = ["bc1_aaa", "bc1_bbb", "bc1_ccc", "bc1_ddd"];
let owners    = ["Exchange", "Miner Pool", "Cold Storage", "Unknown"];

let wanted = "bc1_bbb";
let position = findAddress(addresses, wanted);

if (position > 0) {
  let owner = owners[position - 1];   // pseudocode 1, JavaScript 0
  console.log("Found at position " + position);
  console.log("Owner: " + owner);
} else {
  console.log("Address not tracked");
}
```

### Python

```python
# CODE_RUNNER: Look up bc1_ccc instead.

def find_address(address_list, target):
    index = 1

    for address in address_list:
        if address == target:
            return index
        index = index + 1

    return -1

addresses = ["bc1_aaa", "bc1_bbb", "bc1_ccc", "bc1_ddd"]
owners    = ["Exchange", "Miner Pool", "Cold Storage", "Unknown"]

wanted = "bc1_bbb"
position = find_address(addresses, wanted)

if position > 0:
    owner = owners[position - 1]      # pseudocode 1, Python 0
    print("Found at position " + str(position))
    print("Owner: " + owner)
else:
    print("Address not tracked")
```

## 12. Parallel Lists, Return the Value

Same search, but the procedure returns the fee itself. The caller never sees an index. Better abstraction than topic 11.

### JavaScript

```python
%%js

// CODE_RUNNER: Add a third list for confirmations and return that instead.

function findFee(txids, fees, target) {
  let index = 1;

  for (let txid of txids) {
    if (txid === target) {
      return fees[index - 1];         // pseudocode 1, JavaScript 0
    }
    index = index + 1;
  }

  return -1;
}

let txids = ["tx_9f2a", "tx_4c81", "tx_7de0", "tx_1b55"];
let fees  = [1200, 3400, 850, 2100];

let wanted = "tx_4c81";
let result = findFee(txids, fees, wanted);

if (result > 0) {
  console.log(wanted + " paid " + result + " sats");
} else {
  console.log("Transaction not found");
}
```

### Python

```python
# CODE_RUNNER: Add a third list for confirmations and return that instead.

def find_fee(txids, fees, target):
    index = 1

    for txid in txids:
        if txid == target:
            return fees[index - 1]     # pseudocode 1, Python 0
        index = index + 1

    return -1

txids = ["tx_9f2a", "tx_4c81", "tx_7de0", "tx_1b55"]
fees  = [1200, 3400, 850, 2100]

wanted = "tx_4c81"
result = find_fee(txids, fees, wanted)

if result > 0:
    print(wanted + " paid " + str(result) + " sats")
else:
    print("Transaction not found")
```

## Cheat Sheet

| Idea | Pseudocode | JavaScript | Python |
|------|-----------|------------|--------|
| Assign | `x <- 5` | `let x = 5;` | `x = 5` |
| Output | `DISPLAY(x)` | `console.log(x)` | `print(x)` |
| Input | `INPUT("m")` | `prompt("m")` | `input("m")` |
| First position | `1` | `0` | `0` |
| Procedure | `PROCEDURE f(a)` | `function f(a) {}` | `def f(a):` |
| Blocks | `{ }` | `{ }` | indentation |
| Equal | `=` | `===` | `==` |
| Not equal | `NOT (a = b)` | `a !== b` | `a != b` |
| Append | `APPEND(l, v)` | `l.push(v)` | `l.append(v)` |
| Insert at i | `INSERT(l, i, v)` | `l.splice(i-1, 0, v)` | `l.insert(i-1, v)` |
| Remove at i | `REMOVE(l, i)` | `l.splice(i-1, 1)` | `l.pop(i-1)` |
| Length | `LENGTH(l)` | `l.length` | `len(l)` |
| For each | `FOR EACH x IN l` | `for (let x of l)` | `for x in l:` |
| Repeat until | `REPEAT UNTIL (c)` | `while (!c) {}` | `while not c:` |
| Number to text | automatic | automatic | `str(x)` |
| Printing a list | `DISPLAY(l)` | `a,b` | `['a', 'b']` |

Three gotchas:

1. Python needs `str()` to join a number onto text. JavaScript does not.
2. Python has no braces. Indentation is the block.
3. The two languages print lists differently. Run topic 10 in both.

---## 8. Algorithm, All Three PiecesLoop the fees, guard against an empty block, then judge congestion.

### JavaScript

```python
%%js// CODE_RUNNER: Empty the list. The guard should return 0, not crash.function averageFee(fees) {  let total = 0;  let count = 0;  for (let fee of fees) {    total = total + fee;    count = count + 1;  }  if (count > 0) {    return total / count;  } else {    return 0;  }}let blockFees = [1200, 3400, 850, 2100, 1450];let result = averageFee(blockFees);console.log("Fees: " + blockFees);console.log("Average fee: " + result + " sats");if (result >= 2000) {  console.log("Network: busy");} else {  console.log("Network: calm");}
```

### Python

```python
# CODE_RUNNER: Empty the list. The guard should return 0, not crash.def average_fee(fees):    total = 0    count = 0    for fee in fees:        total = total + fee        count = count + 1    if count > 0:        return total / count    else:        return 0block_fees = [1200, 3400, 850, 2100, 1450]result = average_fee(block_fees)print("Fees: " + str(block_fees))print("Average fee: " + str(result) + " sats")if result >= 2000:    print("Network: busy")else:    print("Network: calm")
```

---## 9. Linear SearchLook for a transaction id inside a block. Missing means -1.

### JavaScript

```python
%%js// CODE_RUNNER: Search for a txid that is not there.function findTx(block, target) {  let index = 1;  for (let txid of block) {    if (txid === target) {      return index;    }    index = index + 1;  }  return -1;}let block = ["tx_9f2a", "tx_4c81", "tx_7de0", "tx_1b55"];let wanted = "tx_7de0";let position = findTx(block, wanted);if (position > 0) {  console.log(wanted + " is at position " + position);} else {  console.log("Not in this block");}
```

### Python

```python
# CODE_RUNNER: Search for a txid that is not there.def find_tx(block, target):    index = 1    for txid in block:        if txid == target:            return index        index = index + 1    return -1block = ["tx_9f2a", "tx_4c81", "tx_7de0", "tx_1b55"]wanted = "tx_7de0"position = find_tx(block, wanted)if position > 0:    print(wanted + " is at position " + str(position))else:    print("Not in this block")
```

---## Cheat Sheet| Idea | Pseudocode | JavaScript | Python ||------|-----------|------------|--------|| Assign | `x <- 5` | `let x = 5;` | `x = 5` || Output | `DISPLAY(x)` | `console.log(x)` | `print(x)` || Input | `INPUT("m")` | `prompt("m")` | `input("m")` || First position | `1` | `0` | `0` || Procedure | `PROCEDURE f(a)` | `function f(a) {}` | `def f(a):` || Blocks | `{ }` | `{ }` | indentation || Equal | `=` | `===` | `==` || Not equal | `NOT (a = b)` | `a !== b` | `a != b` || Append | `APPEND(l, v)` | `l.push(v)` | `l.append(v)` || Insert at i | `INSERT(l, i, v)` | `l.splice(i-1, 0, v)` | `l.insert(i-1, v)` || Remove at i | `REMOVE(l, i)` | `l.splice(i-1, 1)` | `l.pop(i-1)` || Length | `LENGTH(l)` | `l.length` | `len(l)` || For each | `FOR EACH x IN l` | `for (let x of l)` | `for x in l:` || Repeat until | `REPEAT UNTIL (c)` | `while (!c) {}` | `while not c:` || Number to text | automatic | automatic | `str(x)` || Printing a list | `DISPLAY(l)` | `a,b` | `['a', 'b']` |Three gotchas:1. Python needs `str()` to join a number onto text. JavaScript does not.2. Python has no braces. Indentation is the block.3. The two languages print lists differently. Run topic 10 in both.
